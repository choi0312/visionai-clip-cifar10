# CLIP Zero-shot classification

Colab 메뉴에서 GPU 런타임을 선택합니다. 사전학습 모델 추론만 수행하며 추가 학습은 없습니다. 공개 모델이므로 토큰을 요구하지 않습니다.


```python
from pathlib import Path
import os, subprocess, sys
repo = Path("/content/visionai-clip-cifar10")
if not repo.exists():
    subprocess.run(["git", "clone", "https://github.com/choi0312/visionai-clip-cifar10.git", str(repo)], check=True)
os.chdir(repo)
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"], check=True)
```

CUDA PyTorch는 Colab 제공 환경을 사용합니다. 아래는 별도 Python 프로세스로 실행하므로 패키지 설치 전 메모리에 남은 Transformers import의 영향을 받지 않습니다. 원본 CIFAR-10과 공개 체크포인트는 최초 실행에서 다운로드합니다.


```python
os.environ["VISIONAI_ROOT"] = str(repo)
subprocess.run([sys.executable, "experiments/clip_inference.py"], check=True)
subprocess.run([sys.executable, "experiments/analyze_and_plot.py"], check=True)
subprocess.run([sys.executable, "experiments/verify_results.py"], check=True)
```

```python
import json
from IPython.display import display, Image
print(json.dumps(json.loads(Path("results/metrics.json").read_text()), indent=2))
for name in ["predictions.png", "prompt_comparison.png", "confusion.png", "candidate_omission.png"]:
    display(Image(filename=str(Path("figures") / name)))
```
